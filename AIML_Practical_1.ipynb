```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
data = {
    'Study_Hours': [2, 3, 4, 5, 6, 7, 8, 9, 10],
    'Attendance': [60, 65, 70, 75, 80, 82, 85, 90, 95],
    'Marks': [45, 50, 55, 60, 65, 70, 75, 82, 90]
}

df = pd.DataFrame(data)

df
```

```python
correlation = df.corr()

print("Correlation Matrix:")
print(correlation)
```

```python
plt.figure(figsize=(8, 5))

sns.heatmap(correlation, annot=True, cmap='coolwarm', fmt='.2f')

plt.title("Correlation Heatmap")
plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.scatterplot(x='Study_Hours', y='Marks', data=df)

plt.xlabel("Study Hours")
plt.ylabel("Marks")
plt.title("Association between Study Hours and Marks")

plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.scatterplot(x='Attendance', y='Marks', data=df)

plt.xlabel("Attendance")
plt.ylabel("Marks")
plt.title("Association between Attendance and Marks")

plt.show()
```

```python
print("Association with Marks:")
print("Study Hours vs Marks:", round(df['Study_Hours'].corr(df['Marks']), 3))
print("Attendance vs Marks:", round(df['Attendance'].corr(df['Marks']), 3))
```
